```python

import pandas as pd
import json

# Carregar CSV
transacoes = pd.read_csv("data/transacoes.csv")
historico = pd.read_csv("data/historico_atendimento.csv")

# Carregar JSON
with open("data/perfil_financeiro.json", encoding="utf-8") as arquivo:
    perfil = json.load(arquivo)

with open("data/orcamento_mensal.json", encoding="utf-8") as arquivo:
    orcamento = json.load(arquivo)

# Converter datas
transacoes["data"] = pd.to_datetime(transacoes["data"])

print("Todos os arquivos foram carregados!")

display(transacoes.head())
```

```python

# Separar receitas e despesas
receitas = transacoes[
    transacoes["tipo"] == "receita"
]["valor"].sum()

despesas = transacoes[
    transacoes["tipo"] == "despesa"
]["valor"].sum()

saldo = receitas - despesas

print("========== RESUMO FINANCEIRO ==========")

print(f"Receitas: R$ {receitas:.2f}")
print(f"Despesas: R$ {despesas:.2f}")
print(f"Saldo disponível: R$ {saldo:.2f}")

print("=======================================")
```

```python

# Filtrar despesas
despesas_df = transacoes[
    transacoes["tipo"] == "despesa"
]

# Agrupar por categoria
gastos_categoria = (
    despesas_df.groupby("categoria")["valor"]
    .sum()
    .sort_values(ascending=False)
)

print("GASTOS POR CATEGORIA")
display(gastos_categoria)
```

```python

import matplotlib.pyplot as plt

gastos_categoria.plot(
    kind="bar",
    color="steelblue",
    figsize=(10, 5)
)

plt.title("CustoCerto AI - Despesas por Categoria")
plt.xlabel("Categorias")
plt.ylabel("Valor em R$")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python

limites = orcamento["limites_por_categoria"]

print("======= ANÁLISE DO ORÇAMENTO =======")

for categoria, limite in limites.items():

    gasto = gastos_categoria.get(categoria, 0)

    percentual = (gasto / limite) * 100 if limite > 0 else 0

    print(f"\nCategoria: {categoria}")
    print(f"Limite: R$ {limite:.2f}")
    print(f"Gasto: R$ {gasto:.2f}")
    print(f"Utilizado: {percentual:.1f}%")

    if gasto > limite:
        print("ALERTA: Orçamento ultrapassado!")

    elif percentual >= 80:
        print("ATENÇÃO: Próximo do limite!")

    else:
        print("Situação dentro do orçamento.")
```

```python

def simular_economia(meta):

    saldo_atual = receitas - despesas

    print("====== SIMULAÇÃO DE ECONOMIA ======")
    print(f"Meta desejada: R$ {meta:.2f}")
    print(f"Saldo atual: R$ {saldo_atual:.2f}")

    if saldo_atual >= meta:
        print("Parabéns! Sua meta já foi alcançada.")
    else:
        falta = meta - saldo_atual
        print(f"Ainda faltam R$ {falta:.2f} para atingir a meta.")

    print("===================================")

simular_economia(1000)
```

```python

def responder(pergunta):

    pergunta = pergunta.lower()

    # Consulta de receitas
    if "receita" in pergunta or "renda" in pergunta:
        return f"Suas receitas registradas são de R$ {receitas:.2f}."

    # Consulta de saldo
    elif "saldo" in pergunta:
        return f"Seu saldo disponível é de R$ {saldo:.2f}."

    # Consulta de despesas
    elif "gastei" in pergunta or "despesas" in pergunta:
        return f"Suas despesas registradas totalizam R$ {despesas:.2f}."

    # Consulta por categoria
    elif "alimentacao" in pergunta or "alimentação" in pergunta:
        valor = gastos_categoria.get("Alimentacao", 0)
        return f"Seus gastos com alimentação são de R$ {valor:.2f}."

    elif "transporte" in pergunta:
        valor = gastos_categoria.get("Transporte", 0)
        return f"Seus gastos com transporte são de R$ {valor:.2f}."

    elif "moradia" in pergunta:
        valor = gastos_categoria.get("Moradia", 0)
        return f"Seus gastos com moradia são de R$ {valor:.2f}."

    # Consulta de maior gasto
    elif "maior gasto" in pergunta or "gastando mais" in pergunta:
        categoria = gastos_categoria.idxmax()
        valor = gastos_categoria.max()

        return (
            f"Sua maior categoria de despesas é {categoria}, "
            f"com R$ {valor:.2f}."
        )

    # Consulta de orçamento
    elif "orçamento" in pergunta or "orcamento" in pergunta:
        categoria = "Alimentacao"
        gasto = gastos_categoria.get(categoria, 0)
        limite = limites[categoria]

        diferenca = gasto - limite

        if diferenca > 0:
            return (
                f"Alimentação ultrapassou o orçamento em "
                f"R$ {diferenca:.2f}."
            )

        return (
            f"Alimentação está dentro do orçamento. "
            f"Restam R$ {abs(diferenca):.2f}."
        )

    # Meta de economia
    elif "economizar" in pergunta or "economia" in pergunta:
        meta = perfil["objetivos"]["meta_economia_mensal"]
        falta = max(0, meta - saldo)

        return (
            f"Sua meta de economia é R$ {meta:.2f}. "
            f"Seu saldo registrado é R$ {saldo:.2f}. "
            f"Faltam R$ {falta:.2f} para alcançar essa meta."
        )

    else:
        return (
            "Não consegui identificar sua solicitação. "
            "Posso ajudar com despesas, receitas, saldo, "
            "orçamento e metas de economia."
        )


print(responder("Quanto gastei com alimentação?"))
```

```python

print("====================================")
print("       🤖 CUSTOCERTO AI")
print("  Seu assistente de custos mensais")
print("====================================")

print("Digite 'sair' para encerrar a conversa.")

while True:

    pergunta = input("\nVocê: ")

    if pergunta.lower() == "sair":
        print("\nCustoCerto AI: Obrigado por utilizar nosso agente!")
        break

    resposta = responder(pergunta)

    print("\nCustoCerto AI:", resposta)
```
